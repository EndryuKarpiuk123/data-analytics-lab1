## Крок 4. Зчитування CSV-файлу у DataFrame

```python
PATH = '/content/drive/MyDrive/sales.csv'

# Якщо файлу sales.csv немає на Диску - створюємо навчальний набір про продажі кави
import os
if not os.path.exists(PATH):
    rng = np.random.default_rng(42)
    n = 60
    drinks = {"Latte": 38, "Americano": 30, "Cappuccino": 35, "Espresso": 28, "Hot Chocolate": 36}
    names = rng.choice(list(drinks), n)
    cash = rng.choice(["card", "cash"], n, p=[0.8, 0.2])
    dt = pd.to_datetime("2024-03-01 08:00") + pd.to_timedelta(np.sort(rng.integers(0, 60*24*20, n)), unit="m")
    pd.DataFrame({
        "date": dt.date,
        "datetime": dt,
        "cash_type": cash,
        "card": [f"ANON-{rng.integers(1000,9999)}" if c == "card" else np.nan for c in cash],
        "money": [drinks[x] + rng.choice([0, 2, 4]) for x in names],
        "coffee_name": names,
    }).to_csv(PATH, index=False)
    print("Створено навчальний файл:", PATH)

df = pd.read_csv(PATH)  # зчитує CSV у DataFrame
```

## Крок 6. Базові операції з масивами NumPy

```python
arr = np.array([10, 20, 30, 40, 50])      # одновимірний масив зі списку
print("Масив:", arr)
print("Тип елементів:", arr.dtype)         # int64
print("Форма масиву:", arr.shape)          # (5,)

arr2 = arr * 2                             # поелементне множення (векторизація)
print("Помножений масив:", arr2)

arr3 = arr + 5                             # поелементне додавання
print("Масив після додавання 5:", arr3)

# агрегатні функції
print("Сума:", arr.sum())
print("Середнє:", arr.mean())
print("Мінімум:", arr.min())
print("Максимум:", arr.max())
```
