## Завдання 7. Читання CSV з Google Диска та `info()`
Монтуємо Диск (потрібен Google Colab) і зчитуємо `sales.csv`.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
PATH = '/content/drive/MyDrive/sales.csv'

# Якщо файлу sales.csv немає на Диску - створюємо навчальний набір про продажі кави
import os
if not os.path.exists(PATH):
    rng = np.random.default_rng(42)
    n = 60
    drinks = {"Latte": 38, "Americano": 30, "Cappuccino": 35, "Espresso": 28, "Hot Chocolate": 36}
    names = rng.choice(list(drinks), n)
    cash = rng.choice(["card", "cash"], n, p=[0.8, 0.2])
    dt = pd.to_datetime("2024-03-01 08:00") + pd.to_timedelta(np.sort(rng.integers(0, 60*24*20, n)), unit="m")
    pd.DataFrame({
        "date": dt.date,
        "datetime": dt,
        "cash_type": cash,
        "card": [f"ANON-{rng.integers(1000,9999)}" if c == "card" else np.nan for c in cash],
        "money": [drinks[x] + rng.choice([0, 2, 4]) for x in names],
        "coffee_name": names,
    }).to_csv(PATH, index=False)
    print("Створено навчальний файл:", PATH)

df = pd.read_csv(PATH)
df.info()
```

```python
print("Типи тестового набору (завдання 2):")
print(df_test.dtypes)
print()
print("Типи набору з Диска:")
print(df.dtypes)
print()
print(df.dtypes.value_counts())
```

**Висновок:** у наборі з Диска переважають текстові колонки (`object`: дата, час, тип оплати, номер картки, назва напою), числових лише одна - `money` (`float64`/`int64`). У тестовому наборі з завдання 2 числові колонки (`int64`) і текстові (`object`) розподілені приблизно порівну. Pandas не розпізнає дати автоматично - вони читаються як `object`, для роботи з ними їх треба перетворити через `pd.to_datetime()`.
